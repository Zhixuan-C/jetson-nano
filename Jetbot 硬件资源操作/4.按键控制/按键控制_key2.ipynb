## 1.导入库

```python
import RPi.GPIO as GPIO
from time import sleep
```

## 2. 配置管脚

```python
# 配置按钮和蜂鸣器引脚
makerobo_button_pin = 19 # 按键key2
makerobo_buzzer_pin = 6 # 蜂鸣器
# 暂时忽略警告
GPIO.setwarnings(False)
# 使用实际的PIN管脚编码
GPIO.setmode(GPIO.BCM)


# 设置按钮引脚按键管脚为输入和蜂鸣器引脚作为输出
GPIO.setup(makerobo_button_pin, GPIO.IN, pull_up_down=GPIO.PUD_UP)
# 将蜂鸣器引脚设置为输出引脚，并将初始值设置为LOW（关闭）
GPIO.setup(makerobo_buzzer_pin, GPIO.OUT,initial=GPIO.LOW)
```

## 3.执行检测

```python
while True:
    GPIO.setup(makerobo_button_pin, GPIO.IN)
    ver = GPIO.input(makerobo_button_pin)
    GPIO.setup(makerobo_button_pin, GPIO.OUT,initial=GPIO.HIGH)
    # 检查按键是否按下
    if(ver ==0):
        GPIO.output(makerobo_buzzer_pin, GPIO.HIGH)
    else:
        # 没有按键按下, 关闭蜂鸣器
        GPIO.output(makerobo_buzzer_pin, GPIO.LOW)
```
