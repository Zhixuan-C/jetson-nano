## 1. 载入库

```python
from jetbot import Robot
import time
#import Adafruit_PCA9685
#pwm = Adafruit_PCA9685.PCA9685()

# 实例化机器人对象
robot = Robot()
```

## 2.载入创乐博云台函数
makerobo_servo(basechannel,basevalue,tiltchannel,tiltvalue)
> basechannel 定义底座舵机序号 <br>
> basevalue   定义底座舵机旋转角度，单位为度<br>
> tiltchannel 定义倾斜舵机标号<br>
> tiltvalue   定义倾斜舵机角度，单位为度<br>

```python

robot.makerobo_servo(0,90,1,90)
```

```python
robot.set_motors(0.3,0.3,90,60)
```

```python
robot.set_motors(0.5,0.5,90,60)
```

```python
robot.set_motors(0.15,0.15,90,60)
```

```python
robot.forward(1,90,60)#,90,90)
```

```python
robot.forward(0.3)
```

```python
robot.stop()
```

```python
# 舵机云台旋转角度测试
for i in range (30, 120, 5):
    robot.makerobo_servo(0,i,1,i)
    # 舵机反应延时，适当延时
    time.sleep(0.1)
```
