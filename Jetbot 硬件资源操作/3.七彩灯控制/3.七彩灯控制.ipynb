## 1. 导入GPIO控制库

```python
# 导入树莓Pi GPIO库
import RPi.GPIO as GPIO
# 从time模块导入sleep函数
from time import sleep
```

## 2. 定义管脚

```python
# 定义七彩灯控制管脚
makerobo_Led_pin = 25
# 暂时忽略警告
GPIO.setwarnings(False)
# 设置GPIO模式作为 GPIO.BCM
GPIO.setmode(GPIO.BCM)
# 将蜂七彩LED引脚作为输出引脚，并将初始值设置为HIGH（打开）
GPIO.setup(makerobo_Led_pin, GPIO.OUT,initial=GPIO.HIGH)
```

## 3.控制七彩灯

```python
# 关闭七彩灯
GPIO.output(makerobo_Led_pin, GPIO.LOW)
```

```python
GPIO.output(makerobo_Led_pin, GPIO.HIGH)
```
