```python
import RPi.GPIO as GPIO
import time




makerobo_R = 17
makerobo_G = 18
makerobo_B = 27


# 初始化程序
def makerobo_setup(Rpin, Gpin, Bpin):
	global pins
	global p_R, p_G, p_B
	GPIO.setmode(GPIO.BCM)        # 采用实际的物理管脚给GPIO口
	pins = {'pin_R': Rpin, 'pin_G': Gpin, 'pin_B': Bpin}
	GPIO.setwarnings(False)         # 去除GPIO口警告
	for i in pins:
		GPIO.setup(pins[i], GPIO.OUT)   # 设置Pin模式为输出模式
		GPIO.output(pins[i], GPIO.LOW)  # 设置Pin管脚为低电平(0V)关闭LED


# 关闭RGB-LED灯
def makerobo_off():
	for i in pins:
        
		GPIO.output(pins[i], GPIO.LOW)    #  设置Pin管脚为低电平(0V)关闭LED

def makerobo_destroy():
	makerobo_off()  # 关闭RGB-LED灯
	GPIO.cleanup()  # 释放资源

makerobo_setup(makerobo_R, makerobo_G, makerobo_B) # 初始化设置函数
```

## 亮红色灯

```python
GPIO.output(makerobo_R,GPIO.HIGH)
GPIO.output(makerobo_G,GPIO.LOW)
GPIO.output(makerobo_B,GPIO.LOW)
```

## 亮绿色灯

```python
GPIO.output(makerobo_R,GPIO.LOW)
GPIO.output(makerobo_G,GPIO.HIGH)
GPIO.output(makerobo_B,GPIO.LOW)
```

# 亮蓝色灯

```python
GPIO.output(makerobo_R,GPIO.LOW)
GPIO.output(makerobo_G,GPIO.LOW)
GPIO.output(makerobo_B,GPIO.HIGH)
```

## 全亮

```python
GPIO.output(makerobo_R,GPIO.HIGH)
GPIO.output(makerobo_G,GPIO.HIGH)
GPIO.output(makerobo_B,GPIO.HIGH)
```

# 关闭所有的灯

```python
makerobo_off()
```
